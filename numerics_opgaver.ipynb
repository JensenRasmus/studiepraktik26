### Delopgave 2: Lorenzligningerne 

Lorenzligningerne er et ikke-lineært system af 3 differentialligninger
$$
\begin{aligned}
    x'&=\sigma(-x+y)\\
    y'&=rx-y-xz\\
    z'&=-bz+xy,
\end{aligned}
$$
hvor $\sigma,b,r\in \R$ er konstanter. Vi sætter $\sigma = 10$, $b=8/3$ og $r=28$.

**Indsæt 'fy' og 'fz' korrekt i den nedenstående funktion 'fLorenz'**


````python
def fLorenz(x,y,z):
    sigma = 10
    b = 8/3
    r = 28

    fx = sigma*(-x+y)
    
    
    return fx, fy, fz
````

Lorenzligningerne har ikke en lukket løsning, og derfor benytter vi os af Eulers metode, som generelt er givet ved
$$
    y(t_n) = y(t_{n-1})+hf(t_{n-1},y(t_{n-1})),
$$
hvor $h>0$ er skridtstørelsen og $y(t_0) = y_0$ er begyndelsesvædien. 

**Lav koden færdig for y og z koordinatet**

````python
t0 = 0
tEnd = 20
h = 0.0001

Y = np.zeros((3,int(tEnd/h)+1))

Y[0,0] = 5
Y[1,0] = 5
Y[2,0] = 5


for i in range(1,np.shape(Y)[1]):
    fx,fy,fz = fLorenz(Y[0,i-1],Y[1,i-1],Y[2,i-1])

    # Anvend Euler i hvert koordinat 
    Y[0,i] = Y[0,i-1] + h * fx # x
    #Din kode her!
    # y
    # z

    t0 = t0 + h
````

````python
plt.plot(Y[0,:],Y[2,:])
plt.show()
````

Hvis du vælger at gøre $h$ lidt større eller lidt mindre vil du se at billedet ændrer sig. Det skyldes at Lorenzligningerne er meget sensitive over for selv små ændringer i deres numeriske løsning, samt over for ændringer i deres begyndelsesværdi.
